## 함수 정의

```python
def mul(*values):
    result = 1
    # pass
    # print(values)
    # print(type(values))
    for val in values:
        result = result * val
    # print(result)
    return result
    print("-"*20)
```

```python
# 정의부
def comMul(num1, num2):
    # pass
    # 1 ~ 100까지 공배수를 구하기
    # 공배수 : 2 * 3 의 배수
    # 1~100의 6의 배수구하기
    # 6 * 1
    # 6 * 2
    # 6 * 3
    num = num1 * num2
    results = []  # 공배수 전체를 모으는 리스트
    result = 0    # 찾은 공배수
    n = 1         # 1씩 증가시키면서 공배수 찾기 위함
    while result <= 100:
        # num과 result를 1씩 증가 시켜서 결과가 100보다 같거나 작을때까지 반복
        result = num * n
        if result < 100:
            results.append(result)
            n = n + 1
            
        else:
            break
    # print(results)
    return results


    
```

```python
# input_data = input("두 수 입력 예) 2, 3 > ")
input_data
```

```python
num1, num2 = input_data.split(",")
num1, num2 = int(num1), int(num2)
num1, num2
```

```python
# 호출부
results = comMul(num1, num2)
```

```python
print(f"{num1}, {num2}의 1~100의 공배수는 {results}")
```
