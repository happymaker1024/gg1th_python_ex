# 반복문 - for

## for + list

```python
nums = [273, 310, 4, 6, 200, 100, 20, 50, 40, 101]
len(nums)
```

```python
# 100이상의 숫자만 출력
for num in nums:
    # 100이상 존건 참인것만 출력
    if num >= 100:
        print(f"- 100이상의 수: {num}")
        
```

```python
nums
```

```python
# 100이상의 숫자, 100미만인 숫자를 따로 새로운 리스트에 넣어서 각각 리스트로 출력
lower100 = []
higher100 = []
for num in nums:
    # 100이상 존건 참인것만 출력
    if num >= 100:
        # print(f"- 100이상의 수: {num}")
        # 100이상 리스트 추가
        higher100.append(num)
    else:
        # 100미만 리스트 추가
        lower100.append(num)

print(f"- 100이상의 수: {higher100}")
print(f"- 100이상의 수: {lower100}")
```

## for + dict

```python
# nums 의 숫자가 각각 몇번 반복하는지 dict로 만들어서 출력
# {1:횟수 , 2:횟수, 3:횟수, 4:획수, ... }
nums = [1,2,6,7,4,3,2,1,5,9,4,9,7,2,1,3,4,5,6,7,5,6,7,5]
```

```python
counter = {}
counter[2] = 2
counter[3] = 3
counter[4] = 4
counter[3] = 3*2
counter
```

```python
counter = {}

for num in nums:
    # 키값 판단
    if num not in counter :
    # 그 숫자가 제일 처음 나왔을 때, 키값 + 카운트(초기화)
        counter[num] = 1

    else:
    # 그 숫자가 제일 처음 안나왔을때, 카운트
        counter[num] = counter[num] + 1

print(counter)
```

## 리스트 컴프리헨션

```python
array = [i * i for i in range(0, 20, 2)]
array

# 예측 결과 : 
```

```python
# array = [i * i for i in range(0, 20, 2)]
# for 문으로 만들기
array = []
for i in range(0, 20, 2):
    array.append(i * i)

array
```

### 리스트 컴프리헨션의 조건문

```python
# 짝수만 출력하기
even_nums = [] 
for i in range(1, 20+1):
    # i가 짝수인지 판단, 짝수이면, 리스트에 추가
    if i % 2 == 0:
        even_nums.append(i)
even_nums
```

```python
array = [i for i in range(1, 20+1) if i % 2 == 0]
array
```
